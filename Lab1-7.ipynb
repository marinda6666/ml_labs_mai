3. Правда ли, что люди с высоким уровнем стресса имеют более высокое кровяное давление, чем те, кто не подвержен стрессу

````python
# да
data[['Systolic', 'Diastolic']] = data['Blood Pressure (s/d)'].str.split('/', expand=True).astype(float)

high_stress = data[data['Stress Levels'] > data['Stress Levels'].median()]
low_stress = data[data['Stress Levels'] <= data['Stress Levels'].median()]

print('High stress, Systolic:', high_stress['Systolic'].mean())
print('Low stress, Systolic:', low_stress['Systolic'].mean())
print('High stress, Diastolic:', high_stress['Diastolic'].mean())
print('Low stress, Diastolic:', low_stress['Diastolic'].mean())
````

6. Влияет ли физическая активность на вес в возрасте более 50 лет?

````python
over_50 = data[data['Age (years)'] > 50]
weight_by_activity = over_50.groupby('Physical Activity Level')['Weight (kg)'].mean()

max_weight = weight_by_activity.max()
min_weight = weight_by_activity.min()
weight_difference = max_weight - min_weight

print("Yes" if weight_difference > 5 else "No")
print(f"Weight differene: {weight_difference}")
````

7. Сравните минимальный вес мужчин с высоким и низким заработком

8. Правда ли, что у людей с индексом массы тела больше 75го перцентиля встречаются хронические заболевания чаще, чем у людей с индесом массы тела в переделах от 30 до 60 перцентиля?

````python
p75 = data['BMI'].quantile(q=0.75)
p60 = data['BMI'].quantile(q=0.6)
p30 = data['BMI'].quantile(q=0.3)

high_bmi_group = data[data['BMI'] > p75]
mid_bmi_group = data[(data['BMI'] >= p30) & (data['BMI'] <= p60)]

high_bmi_chronic = high_bmi_group['Chronic Diseases']
mid_bmi_chronic = mid_bmi_group['Chronic Diseases']

high_bmi_rate = high_bmi_chronic.dropna().shape[0] / high_bmi_chronic.shape[0]
mid_bmi_rate = mid_bmi_chronic.dropna().shape[0] / mid_bmi_chronic.shape[0]

print(f'BMI > 75: {high_bmi_rate:.2f}')
print(f'BMI from 30 to 60: {mid_bmi_rate:.2f}')
````

9. Правда ли, что люди с высоким уровнем дохода имеют больший уровень физической активности, чем люди с низким доходом

````python
high_salary = data[data["Income Level"] == "High"]
low_salary = data[data["Income Level"] == "Low"]

count_high_active = (high_salary["Physical Activity Level"] != "Low").sum()
count_low_active = (low_salary["Physical Activity Level"] != "Low").sum()

print("yes" if count_high_active > count_low_active else "no")
````

10. Какая доля мужчин с высокой частотой пребывания на солнце имеет высокий заработок?

````python
sun_exposure_median = data["Sun Exposure"].median()
high_sun_exposure = data["Sun Exposure"] > sun_exposure_median

men_high_sun = data[(data["Gender"] == "Male") & high_sun_exposure]

high_income_men = men_high_sun[men_high_sun["Income Level"] == "High"]
proportion = len(high_income_men) / len(men_high_sun)

print(f"Proportion: {proportion:.3f}")
````
